## 8
In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()` method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as output

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values. 

```python
# Calculate AQI rolling average over 5 days
rolling_average = aqi_sb['aqi'].rolling(window='5D').mean()
rolling_average
```

Without creating any new variables, add the mean of the AQI over a 5-day rolling window as a new column named `five_day_average` to the `aqi_sb` data frame.

Make a line plot showing both the daily AQI and the 5-day average (5-day average on top of the AQI). Make necessary updates for the plot to be accurate, informative, and polished, even if simple. You're welcome to use plotting methods and packages beyond what we have covered in class!

```python

fig, ax = plt.subplots()
ax.set_ylim(top = 325)


# Add bars indicationg AQI safety
good = ax.axhspan(ymin = 0, ymax = 50, color = 'green', alpha = 0.3, label = '_Good')
moderate = ax.axhspan(ymin = 51, ymax =100, color = 'yellow',  alpha = 0.3, label = '_Moderate')
unhealthy1 = ax.axhspan(ymin = 101, ymax = 150, color = 'orange',  alpha = 0.3, label = '_Unhealthy for Sensitive Groups')
unhealthy2 = ax.axhspan(ymin = 151, ymax = 200, color = 'red',  alpha = 0.3, label = '_Unhealthy')
v_unhealthy = ax.axhspan(ymin = 201, ymax = 300, color = 'purple',  alpha = 0.3, label = '_Very Unhealthy')
hazardous = ax.axhspan(ymin = 301, ymax = 400, color = 'maroon',  alpha = 0.3, label = '_Hazardous')
# Create a custom legend for safety levels
safety_legend = ax.legend(handles = [good, moderate, unhealthy1, unhealthy2, v_unhealthy, hazardous], 
                          labels = ['Good', 'Moderate', 'Unhealthy for \nSensitive Groups', 'Unhealthy', 'Very Unhealthy', 'Hazardous'],
                          loc = 'center right', bbox_to_anchor=(1.39, 0.45), 
                          title = "AQI Safety Category",
                          title_fontproperties = {'weight':'bold'})
ax.add_artist(safety_legend)

#Plot the daily AQI and 5-day average
aqi = plt.plot(aqi_sb['aqi'], alpha = 0.5, color = 'black', label = "Daily AQI", linestyle = 'dotted')
five_day = plt.plot(aqi_sb['five_day_average'], color = 'black', label = 'Five-Day Average AQI')
thomas = ax.axvspan(pd.Timestamp("2017-12-04"), 
                    pd.Timestamp("2018-01-12"), 
                    color="gray", 
                    alpha=0.3, 
                    label="Thomas Fire")

# Create a legend for the timeseries data (AQI, 5-day average, and Thomas Fire)
fig.legend(loc = 'outside upper right', 
           bbox_to_anchor=(1.25, 0.85), 
           title = 'AQI Timeseries', 
           title_fontproperties = {'weight':'bold'})

plt.xlabel('Date')
plt.ylabel('Air Quality Index')
plt.title('Air Quality Index (AQI) in Santa Barbara County (2017 - 2018)')
plt.xticks(rotation = 30)
plt.show()
;
```
